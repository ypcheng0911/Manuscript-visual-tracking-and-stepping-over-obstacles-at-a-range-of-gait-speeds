```python
import cv2
import glob

subj_list = ["D02","D03","D04","D05","D06","D07","D08","D09","D10",
             "N01","N02","N03","N04","N05","N06","N07","N08","N09","N10"]

for subj in subj_list:
    filepath = "/Users/YPCheng/Desktop/Eye tracking raw data/"+subj+"/Data/Media"
    txt_path = "/Users/YPCheng/Desktop/Eye tracking raw data/"+subj
    for videopath in glob.glob(filepath+"/*.mp4"):
        cap = cv2.VideoCapture(videopath)
        frame_rate = cap.get(cv2.CAP_PROP_FPS)
        timestamps = []
        frame_no = 0
        while(cap.isOpened()):
            frame_exists, curr_frame = cap.read()
            if frame_exists:
                #print("for frame : " + str(frame_no) + "   timestamp is: ", str(cap.get(cv2.CAP_PROP_POS_MSEC)))
                timestamps.append(cap.get(cv2.CAP_PROP_POS_MSEC))
            else:
                break
            frame_no += 1
        cap.release()
        filename = videopath[len(filepath)+1:-4]

        f = open(txt_path+"/"+filename+".txt","w")
        f.write("Frame rate: "+str(frame_rate))
        f.write("\n")
        f.write("Timestamps:\n")
        for t in timestamps:
            f.write(str(t)+"\n")
        f.close()
```
