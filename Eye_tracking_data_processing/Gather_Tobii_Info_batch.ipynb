```python
import glob
import numpy
subj_list = ["D01","D02","D03","D04","D05","D06","D07","D08","D09","D10"
             ,"N01","N02","N03","N04","N05","N06","N07","N08","N09","N10"]

study_path = "<path_to_raw_data>/Eye tracking raw data/"
for subj in subj_list:
    rec_path = study_path+subj+"/Data/Recordings/*.rec"
    par_path = study_path+subj+"/Data/Participants/*.exp"
    med_path = study_path+subj+"/Data/Media/*.xml"
    txt_path = study_path+subj
    G_rec = glob.glob(rec_path)
    G_par = glob.glob(par_path)
    G_med = glob.glob(med_path)

    # [Process recording info]
    txt_recordingID = []
    txt_participantID = []
    txt_videoID = []
    for temp_path in G_rec:
        # All info
        f = open(temp_path,"r")
        f.readline()
        f.readline()
        f.readline()
        f.readline()
        RecordingID = f.readline()
        f.readline()
        VideoID = f.readline()
        f.readline()
        ParticipantID = f.readline()
        txt_recordingID.append(RecordingID[7:-7])
        txt_participantID.append(ParticipantID[17:-17])
        txt_videoID.append(VideoID[15:-13])
    # Write .txt file
    f = open(txt_path+"/txt_recording.txt","w")
    for idx in range(len(G_rec)):
        RID,PID,VID = txt_recordingID[idx],txt_participantID[idx],txt_videoID[idx]
        f.write(RID+','+PID+','+VID)
        f.write('\n')
    f.close()

    # [Process participant info]
    txt_ParticipantID = []
    txt_CondName = []
    for temp_path in G_par:
        # Participant info
        f = open(temp_path,"r")
        f.readline()
        f.readline()
        ParticipantID = f.readline()
        CondName = f.readline()
        txt_ParticipantID.append(ParticipantID[7:-7])
        txt_CondName.append(CondName[8:-8])
    # Write .txt file
    f = open(txt_path+"/txt_participant.txt","w")
    for idx in range(len(G_par)):
        PID,CN = txt_ParticipantID[idx],txt_CondName[idx]
        f.write(PID+','+CN)
        f.write('\n')
    f.close()

    # [Process video info]
    txt_VideoID = []
    txt_VideoName = []
    for temp_path in G_med:
        # Video info
        f = open(temp_path,"r")
        f.readline()
        f.readline()
        videoID = f.readline()
        f.readline()
        f.readline()
        videoName = f.readline()
        txt_VideoID.append(videoID[7:-7])
        txt_VideoName.append(videoName[20:-20])
        # Testing for writing .txt file
    f = open(txt_path+"/txt_video.txt","w")
    for idx in range(len(G_med)):
        VID,VN = txt_VideoID[idx],txt_VideoName[idx]
        f.write(VID+','+VN)
        f.write('\n')
    f.close()

    # [Save summary file with the sorted condition and video names]
    VN_sorted = []
    VID_sort_idx = numpy.argsort(txt_VideoID)
    for idx in VID_sort_idx.tolist():
        VN_sorted.append(txt_VideoName[idx])
        pID_sorted = []
    vID_sort_idx = numpy.argsort(txt_videoID)
    for idx in vID_sort_idx.tolist():
        pID_sorted.append(txt_participantID[idx])
    # Write .txt file
    f = open(txt_path+"/summary.txt","w")
    for idx in range(len(pID_sorted)):
        f.write(VN_sorted[idx]+','+txt_CondName[txt_ParticipantID.index(pID_sorted[idx])])
        f.write('\n')
    f.close()

    print(subj+' file saving done.')
```
